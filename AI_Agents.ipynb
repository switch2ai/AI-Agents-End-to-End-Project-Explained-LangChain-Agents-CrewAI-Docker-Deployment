```python
from langchain_community.tools import DuckDuckGoSearchRun

search = DuckDuckGoSearchRun()
```

```python
search.run("What are Tamilnadu election results?")
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model_name="gpt-4o")
```

```python
llm.invoke("What are Tamilnadu election results?")
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model=llm,
    tools=[search]
    )
```

```python
from langchain.messages import HumanMessage

message = {
    "messages" : [HumanMessage(content="What are Tamilnadu election results?")]
}
```

```python
output = agent.invoke(message)
```

```python
from langchain.tools import tool

@tool
def add(a:int, b:int) -> int:
  """This is used to add two numbers.
  """
  return a + b
```

```python
>> Docstring : it is a description of tool
>> type hints : Agents will know what input to give to tool
```

```python
@tool("addidtion", description="This is used to add two numbers.")
def add(a:int, b:int) -> int:
  """This is used to add two numbers.
  """
  return a + b
```
